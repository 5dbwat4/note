```python
import time
import numpy as np
from gensim.models import Word2Vec
probe_words = ['金融', '喜欢', '中国', '北京']
PRINT_TOPN = 5    # 打印前 5 个相似词
EVAL_TOPN = 10    # 用前 10 个词计算重合度
def get_neighbors(wv, word, topn):
    """安全地获取相似词，词不在词表中时返回空列表"""
    try:
        return wv.most_similar(word, topn=topn)
    except KeyError:
        return []
base_wv = model.wv
baseline = {w: get_neighbors(base_wv, w, EVAL_TOPN) for w in probe_words}
print("Baseline: vector_size=100, window=5, sg=1, min_count=5")
for w, res in baseline.items():
    print(f"  {w}: {[(x, round(s, 4)) for x, s in res[:PRINT_TOPN]]}")
# 每次只改变一个参数：名称 -> 参数字典
EXPERIMENTS = [
    ("vector_size=50",  dict(vector_size=50,  window=5,  sg=1, min_count=5)),
    ("vector_size=200", dict(vector_size=200, window=5,  sg=1, min_count=5)),
    ("window=2",        dict(vector_size=100, window=2,  sg=1, min_count=5)),
    ("window=10",       dict(vector_size=100, window=10, sg=1, min_count=5)),
    ("sg=0 (CBOW)",     dict(vector_size=100, window=5,  sg=0, min_count=5)),
    ("min_count=1",     dict(vector_size=100, window=5,  sg=1, min_count=1)),
    ("min_count=20",    dict(vector_size=100, window=5,  sg=1, min_count=20)),
]
results = {}
for name, params in EXPERIMENTS:
    print("\n" + "=" * 60)
    print(f"训练配置: {name}  {params}")
    t0 = time.perf_counter()
    m = Word2Vec(corpus_file=corpus_file, workers=cpu_count(),
                 epochs=5, seed=42, **params)   # epochs 固定，保证公平比较
    elapsed = time.perf_counter() - t0
    results[name] = {
        "params": params,
        "vocab": len(m.wv),
        "time": elapsed,
        "neighbors": {w: get_neighbors(m.wv, w, EVAL_TOPN) for w in probe_words},
    }
    print(f"词表大小: {len(m.wv)}  训练耗时: {elapsed:.1f}s")
    for w in probe_words:
        res = results[name]["neighbors"][w]
        print(f"  {w}: {[(x, round(s, 4)) for x, s in res[:PRINT_TOPN]]}")
# 与 baseline 的 top-10 邻词重合度，衡量参数变化对结果的影响
print("\n" + "=" * 60)
print(f"与 baseline 的 top-{EVAL_TOPN} 邻词重合度（Jaccard，越高越稳定）")
for name, r in results.items():
    scores = []
    for w in probe_words:
        a = {x for x, _ in baseline[w]}
        b = {x for x, _ in r["neighbors"][w]}
        scores.append(len(a & b) / len(a | b) if (a | b) else 0.0)
    print(f"  {name:16s} 平均={np.mean(scores):.3f}  "
          f"各词={[round(x, 2) for x in scores]}")
```

```python
# ============================================================
# 相似词与词向量关系的拓展探索
# 包括：分主题相似词、词对相似度、词向量类比、找不同类
# ============================================================
import numpy as np
from gensim.models import KeyedVectors
# 若前面没有加载，则从文件加载
try:
    word2vec_model
except NameError:
    word2vec_model = KeyedVectors.load_word2vec_format("embedding.txt")
def show_similar(word, topn=5):
    """打印一个词的相似词；词不在词表中时给出提示"""
    try:
        res = word2vec_model.most_similar(word, topn=topn)
        print(f"{word}: " + ", ".join(f"{w}({s:.3f})" for w, s in res))
    except KeyError:
        print(f"{word}: 未出现在词表中（可能被 min_count 过滤）")
# 1) 按语义主题分组观察，检查同类词是否相互召回
groups = {
    "城市":       ["北京", "上海", "广州", "深圳", "天津"],
    "国家/地区":  ["中国", "美国", "日本", "法国", "德国"],
    "学科":       ["数学", "物理", "化学", "生物", "历史"],
    "情感/评价":  ["喜欢", "讨厌", "高兴", "悲伤", "优秀"],
}
for group, words in groups.items():
    print("\n" + "=" * 60)
    print(f"【{group}】")
    for w in words:
        show_similar(w, topn=5)
# 2) 指定词对的余弦相似度，观察语义关系强弱
print("\n" + "=" * 60)
print("【词对余弦相似度】")
pairs = [
    ("北京", "上海"), ("北京", "中国"), ("中国", "美国"),
    ("喜欢", "讨厌"), ("喜欢", "高兴"), ("数学", "物理"),
    ("数学", "历史"), ("老师", "教师"),
]
for a, b in pairs:
    try:
        print(f"  sim({a}, {b}) = {word2vec_model.similarity(a, b):.4f}")
    except KeyError as e:
        print(f"  sim({a}, {b}) 跳过：{e}")
# 3) 词向量类比：positive 相加、negative 相减，观察语义关系
print("\n" + "=" * 60)
print("【词向量类比】")
analogies = [
    (["日本", "北京"], ["中国"]),   # 中国:北京 = 日本:?  -> 期望 东京
    (["中国", "东京"], ["日本"]),   # 日本:东京 = 中国:?  -> 期望 北京
    (["女人", "国王"], ["男人"]),   # 男人:国王 = 女人:?  -> 期望 女王
]
for pos, neg in analogies:
    try:
        res = word2vec_model.most_similar(positive=pos, negative=neg, topn=5)
        print(f"  {pos} - {neg}: " + ", ".join(f"{w}({s:.3f})" for w, s in res))
    except KeyError as e:
        print(f"  {pos} - {neg} 跳过：{e}")
# 4) 找不同类：判断哪个词与其余词最不相关
print("\n" + "=" * 60)
print("【找不同类】")
for words in [["北京", "上海", "天津", "苹果"],
              ["数学", "物理", "化学", "诗歌"],
              ["喜欢", "讨厌", "高兴", "桌子"]]:
    try:
        print(f"  {words} -> {word2vec_model.doesnt_match(words)}")
    except KeyError as e:
        print(f"  {words} 跳过：{e}")
# 5) 定量比较：组内平均相似度 vs 组间平均相似度
def mean_sim(words):
    """计算一组词内部两两余弦相似度的平均值（忽略不在词表中的词）"""
    vals = []
    for i in range(len(words)):
        for j in range(i + 1, len(words)):
            try:
                vals.append(word2vec_model.similarity(words[i], words[j]))
            except KeyError:
                pass
    return float(np.mean(vals)) if vals else float("nan")
print("\n" + "=" * 60)
print("【组内平均相似度】")
for group, words in groups.items():
    print(f"  {group}: {mean_sim(words):.4f}")
city_words = [w for w in groups["城市"] if w in word2vec_model.key_to_index]
subject_words = [w for w in groups["学科"] if w in word2vec_model.key_to_index]
cross = [word2vec_model.similarity(a, b)
         for a in city_words for b in subject_words]
print(f"  城市 vs 学科 组间平均相似度: {np.mean(cross):.4f}")
```
